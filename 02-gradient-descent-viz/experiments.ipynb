## Phase 2: contour plots

2×2 figure of all surfaces → `assets/surfaces.png`. Rosenbrock with linear vs. log levels.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
surfaces = [gd.ROUND_BOWL, gd.STRETCHED_BOWL, gd.ROSENBROCK, gd.HIMMELBLAU]
for ax, s in zip(axes.flat, surfaces):
    gd.plot_contours(s, ax, log_levels=(s.name == "rosenbrock surface")) 
fig.tight_layout()
fig.savefig("assets/surfaces.png", dpi=120)
plt.show()
```
